```python
available_seats=['A1','A2','A3','A4','A5','B1','B2','B3','B4','B5']
booked_seats=[]

def showavailableseats():
    print("available_seats:")
    for seat in available_seats:
        if seat in available_seats:
            print(seat,end=' ')
            
        else:
            print(seat,'is booked')


def bookseat(book_seat_number):
    if book_seat_number in available_seats:
        if book_seat_number in booked_seats:
            print("seat already booked")
        else:
            booked_seats.append(book_seat_number)
            available_seats.remove(book_seat_number)
            print("seat booked")
            print(booked_seats)
            
    else:
        print("seat is not available")
         


def cancelbooking(cancel_seat):
    if cancel_seat in booked_seats:
        booked_seats.remove(cancel_seat)
        print("seat cancel successfully")
    else:
        print("seat is not booked")


def viewbookingseat():
    print("booked seat")
    if len(booked_seats)==0:
        print("no booked seats")
    else:
        for seat in booked_seats:
            print(seat,end=' ')

def searchbooking(search_seat):
    if search_seat in booked_seats:
        print(booked_seats,"seat is booked")
    elif search_seat in available_seats:
        print(available_seats,"seat is available")
    else:
        print("invalid seat number")
        
def countavailableseat():
    print("available seat")
    print(len(available_seats))
    


def main():
    while(True):
        print()
        print("1 for show avaliable seats")
        print("2 for booking")
        print("3 for cancel booking")
        print("4 for view book seat")
        print("5 for search booking")
        print("6. for count available seat")
        print("7 for exit")
        choice=int(input("enter your choice"))
        if(choice==1):
            showavailableseats()
        elif(choice==2):
            book_seat_number=input("enter the seat number you want to book").upper()
            bookseat(book_seat_number)
        elif(choice==3):
            cancel_seat=input("Enter your seat number to cancel your booking").upper()
            cancelbooking(cancel_seat)
        elif(choice==4):
            viewbookingseat()
        elif(choice==5):
            search=input("enter the seat number u wnat to search").upper()
            searchbooking(search)
        elif(choice==6):
            countavailableseat()
            
        else:
            print("Thank you")
            break
        
        
main()
```
